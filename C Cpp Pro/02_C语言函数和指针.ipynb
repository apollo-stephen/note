# C语言函数和指针

这是C/C++中最容易混淆的两个概念，核心区别在于**谁是“主体”**：
- **指针函数**：本质是**函数**，返回值是一个指针。
- **函数指针**：本质是**指针**，指向一个函数的入口地址。


### 一、指针函数（返回指针的函数）
#### 1. 定义与语法
指针函数是一个**函数**，其返回值是一个指针（即内存地址）。
```cpp
返回类型* 函数名(参数列表);
```
- **关键点**：`*` 与返回类型结合（如 `int*`），表示函数返回一个“指向 `int` 的指针”。
- **无需括号**：因为 `*` 优先级低于 `()`，`int* p(int)` 会被编译器解析为“返回 `int*` 的函数 `p`”。

#### 2. 示例代码（逐行解析）
```cpp
#include <iostream>

// 指针函数：返回一个 int* 指针
int* createArray(int size, int initValue) {
    // 动态分配数组，返回数组首地址（int* 类型）
    int* arr = new int[size];  // new 返回的就是 int*
    for (int i = 0; i < size; ++i) {
        arr[i] = initValue;     // 初始化数组元素
    }
    return arr;  // 返回数组首地址（指针）
}

int main() {
    // 调用指针函数，接收返回的指针
    int* myArr = createArray(5, 10);  // myArr 指向动态分配的数组
    
    for (int i = 0; i < 5; ++i) {
        std::cout << myArr[i] << " ";  // 输出 10 10 10 10 10
    }
    
    delete[] myArr;  // 释放动态内存（重要！）
    return 0;
}
```

#### 3. 常见坑点（⚠️ 重点）
**禁止返回局部变量的指针**：局部变量在函数返回后会被销毁，返回其地址会导致“野指针”（程序崩溃或未定义行为）。
```cpp
int* badExample() {
    int temp = 100;  // 局部变量，函数返回后内存被回收
    return &temp;     // ❌ 错误：返回局部变量的地址
}
```


### 二、函数指针（指向函数的指针）
#### 1. 定义与语法
函数指针是一个**指针变量**，存储的是函数在内存中的入口地址。
```cpp
返回类型 (*函数指针名)(参数列表);
```
- **关键点**：`*` 与函数指针名必须用括号包裹（如 `(*p)`），强制让 `*` 先与 `p` 结合，表示“p 是一个指针”。
- **如果不加括号**：`int* p(int)` 会变成“返回 `int*` 的函数”（即上面的指针函数）。

#### 2. 简化示例（回顾）
```cpp
#include <iostream>

// 目标函数
int add(int a, int b) { return a + b; }

int main() {
    // 函数指针 p：指向 int(int, int) 类型的函数
    int (*p)(int, int) = add;  // 函数名 add 就是函数的入口地址
    
    std::cout << p(3, 5) << std::endl;  // 等价于 add(3,5)，输出 8
    return 0;
}
```


### 三、核心区别对比表
| 维度         | 指针函数（返回指针的函数）       | 函数指针（指向函数的指针）       |
|--------------|----------------------------------|----------------------------------|
| **本质**     | 是一个**函数**                   | 是一个**指针变量**               |
| **语法**     | `int* p(int);`（* 与返回类型结合） | `int (*p)(int);`（* 与变量名结合，需括号） |
| **用途**     | 返回动态分配的内存、数组/字符串地址等 | 实现回调函数、模拟多态、传递函数逻辑等 |
| **示例**     | `int* createArray(int size);`   | `int (*p)(int, int) = add;`     |


### 四、一句话总结
- 看**括号**：有括号 `(*p)` 的是**函数指针**（p 是指针）；
- 看**主体**：“指针函数”重点在“函数”（返回指针），“函数指针”重点在“指针”（指向函数）。
